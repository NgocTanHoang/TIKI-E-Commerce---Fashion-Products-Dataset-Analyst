# Business Performance & Inventory Management (Tiki Data)

## 1. Importing Libraries and Loading Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import glob
from sklearn.model_selection import train_test_split
```

```python
files = glob.glob("../data/*.csv")

df = pd.concat([pd.read_csv(f, encoding='utf-8') for f in files], ignore_index=True)

print(df.shape)
```

## 2. Data Cleaning and Preprocessing

Drop the 'Unnamed: 0' column as it is not needed for analysis.

```python
df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
print("Name of columns:", df.columns)
```

Checking for null values in the dataset and dropping them if any.

```python
df.isnull().sum().sum()
```

```python
df.dropna(inplace=True)

df.isnull().sum().sum()
```

```python
print(df.isnull().sum())
```

Check duplicate rows in the dataset. If there are any duplicate rows, they can skew the analysis and lead to inaccurate insights. Removing duplicates ensures that each record is unique and contributes to a more accurate analysis of the data.

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates(subset="id")
```

Check original price and price columns for any inconsistencies. The original price should always be greater than or equal to the price, as it represents the initial price before any discounts. If there are rows where the original price is less than the price, it indicates a data inconsistency that needs to be addressed.

```python
df[df["original_price"] < df["price"]]
```
